```python
#转换成txt文件
!pip install openai-whisper -q
import whisper
from google.colab import files
file_name = "入群测试题.mp4"  # ← 改成你上传的文件名
print("⏳ 加载模型中（第一次会下载模型，稍等几分钟）...")
model = whisper.load_model("large")
print("⏳ 正在识别...")
result = model.transcribe(f"/content/{file_name}", language="ja")

print("\n结果：\n")
all_lines = []
for seg in result["segments"]:
    text = seg["text"].strip()
    all_lines.append(text)
    print(text)

output_txt = file_name.rsplit(".", 1)[0] + "转录.txt"
with open(output_txt, "w", encoding="utf-8") as f:
    f.write("\n".join(all_lines))
files.download(output_txt)
print(f"\n 导出：{output_txt}")
```

```python
#转换成ass文件
!pip install openai-whisper -q
import whisper
from google.colab import files
import math

file_name = "MS3EP2.mp3"  # ← 改成你上传的文件名

print("⏳ 加载模型中（第一次会下载模型，稍等几分钟）...")
model = whisper.load_model("large")
print("⏳ 正在识别...")
result = model.transcribe(f"/content/{file_name}", language="ja")

def seconds_to_ass_time(secs):
    h = int(secs // 3600)
    m = int((secs % 3600) // 60)
    s = int(secs % 60)
    cs = int((secs - int(secs)) * 100) 
    return f"{h}:{m:02d}:{s:02d}.{cs:02d}"    # 注意这里是点分隔，也可以换成逗号

ass_header = """[Script Info]
Title: Whisper Transcription
ScriptType: v4.00+
Collisions: Normal
PlayResX: 1920
PlayResY: 1080
Timer: 100.0000

[V4+ Styles]
Format: Name, Fontname, Fontsize, PrimaryColour, SecondaryColour, OutlineColour, BackColour, Bold, Italic, Underline, StrikeOut, ScaleX, ScaleY, Spacing, Angle, BorderStyle, Outline, Shadow, Alignment, MarginL, MarginR, MarginV, Encoding
Style: Default,Microsoft YaHei,28,&H00FFFFFF,&H000000FF,&H00000000,&H00000000,0,0,0,0,100,100,0,0,1,2,2,2,10,10,10,1

[Events]
Format: Layer, Start, End, Style, Name, MarginL, MarginR, MarginV, Effect, Text
"""

ass_events = []
for seg in result["segments"]:
    start = seconds_to_ass_time(seg["start"])
    end = seconds_to_ass_time(seg["end"])
    text = seg["text"].strip()
    ass_events.append(f"Dialogue: 0,{start},{end},Default,,0,0,0,,{text}")

ass_content = ass_header + "\n".join(ass_events)

output_ass = file_name.rsplit(".", 1)[0] + "[自动轴].ass"
with open(output_ass, "w", encoding="utf-8") as f:
    f.write(ass_content)

files.download(output_ass)
print(f"\n✅ 导出字幕文件：{output_ass}")
print("🎯 可以用播放器（如 VLC / PotPlayer）直接加载 .ass 字幕")
```
